Las nueve de Han en la matriz del material; y las 868 series de CAMMESA, cada una con sus ítems.</p></div>

```python
import time
import itertools
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
warnings.filterwarnings("ignore")

AZUL, AMBAR, GRIS, ROJO, VERDE = "#0055a5", "#f0a800", "#6b7280", "#c0392b", "#2e8b57"
pd.set_option("display.width", 140); pd.set_option("display.max_colwidth", 90)
nombre = lambda s: "{" + ", ".join(sorted(s)) + "}"                                            # para imprimir un frozenset

# 1 · Las nueve canastas de Han (tabla 6.1 del material)
han = [["I1", "I2", "I5"], ["I2", "I4"], ["I2", "I3"], ["I1", "I2", "I4"], ["I1", "I3"], ["I2", "I3"], ["I1", "I3"], ["I1", "I2", "I3", "I5"], ["I1", "I2", "I3"]]
codificador = TransactionEncoder()
M_han = pd.DataFrame(codificador.fit(han).transform(han), columns=codificador.columns_, index=[f"T{i + 1}00" for i in range(9)])
print("Las nueve canastas como matriz binaria:"); display(M_han.astype(int))

# 2 · Las canastas de CAMMESA: una por serie, con sus atributos y su comportamiento como ítems
CANDIDATAS = [Path("datos/demanda-ltimos-aos.csv"), Path("demanda-ltimos-aos.csv"), Path("../datos/demanda-ltimos-aos.csv"), Path("../Regresion/datos/demanda-ltimos-aos.csv")]
RUTA = next((p for p in CANDIDATAS if p.exists()), None)
if RUTA is None:
    raise FileNotFoundError(f"No encontré el CSV. Ponelo en alguna de estas rutas: {[str(p) for p in CANDIDATAS]}")
demanda = pd.read_csv(RUTA, encoding="utf-8-sig", dtype={"indice_tiempo": str})
demanda["periodo"] = pd.PeriodIndex(demanda["indice_tiempo"], freq="M")
demanda["serie"] = demanda["agente_nemo"] + " | " + demanda["tarifa"]
demanda = demanda.sort_values(["serie", "periodo"])
def perfil(g):
    """Los seis números de la clase 41."""
    s = g.set_index("periodo")["demanda_MWh"]
    if len(s) < 24 or (s <= 0).mean() > 0.1:
        return None
    pos = s[s > 0]
    cambio = (s / s.shift(1) - 1)[s.shift(1) > 0].clip(-1, 3)
    mes = pos.index.month
    verano, invierno = pos[np.isin(mes, [12, 1, 2])].mean(), pos[np.isin(mes, [6, 7, 8])].mean()
    anios = np.array([(p - pos.index[0]).n for p in pos.index]) / 12
    return pd.Series({"log_tamano": np.log10(pos.mean()), "estacionalidad": np.log2(verano / invierno),
                      "volatilidad": cambio.abs().median(), "tendencia_anual": np.polyfit(anios, np.log10(pos.to_numpy()), 1)[0],
                      "frac_caidas": (cambio < -0.15).mean(), "cv": pos.std() / pos.mean()})
perfiles = demanda.groupby("serie").apply(perfil).dropna()
atributos = demanda.groupby("serie")[["region", "categoria_demanda", "categoria_tarifa"]].first().loc[perfiles.index]

# Los umbrales que convierten números en ítems (una decisión, dicha acá)
VOL_ALTA, VOL_BAJA = perfiles["volatilidad"].quantile([0.75, 0.25])
TAM_ALTO, TAM_BAJO = perfiles["log_tamano"].quantile([0.75, 0.25])
def a_items(serie):
    p, a = perfiles.loc[serie], atributos.loc[serie]
    items = {f"región {a['region'].title()}", "gran usuario" if a["categoria_demanda"] == "Gran Usuario" else "distribuidor", f"tarifa {a['categoria_tarifa'].split('/')[0].lower()}"}
    if p["estacionalidad"] > np.log2(1.2): items.add("pico de verano")                         # 20 % más en verano que en invierno
    elif p["estacionalidad"] < -np.log2(1.2): items.add("pico de invierno")                    # 20 % más en invierno
    if p["tendencia_anual"] > np.log10(1.05): items.add("creciente")                           # crece más de 5 % por año
    elif p["tendencia_anual"] < np.log10(0.95): items.add("decreciente")                       # cae más de 5 % por año
    if p["volatilidad"] > VOL_ALTA: items.add("volátil")                                       # el cuarto más volátil
    elif p["volatilidad"] < VOL_BAJA: items.add("estable")                                     # el cuarto más estable
    if p["frac_caidas"] >= 0.2: items.add("caídas frecuentes")                                 # cae más de 15 % en uno de cada cinco meses o más
    if p["log_tamano"] > TAM_ALTO: items.add("grande")
    elif p["log_tamano"] < TAM_BAJO: items.add("chico")
    return sorted(items)
canastas = [a_items(s) for s in perfiles.index]
M = pd.DataFrame(codificador.fit(canastas).transform(canastas), columns=codificador.columns_, index=perfiles.index)
DESCRIPTIVOS = [c_ for c_ in M.columns if c_.startswith(("región", "tarifa", "gran usuario", "distribuidor")) or c_ in ("grande", "chico")]   # lo que se sabe del agente (el tamaño, también)
COMPORTAMIENTO = [c_ for c_ in M.columns if c_ not in DESCRIPTIVOS]
print(f"Canastas de CAMMESA: {len(M)} series · {M.shape[1]} ítems ({len(DESCRIPTIVOS)} de atributos, {len(COMPORTAMIENTO)} de comportamiento) · tamaño medio {M.sum(axis=1).mean():.2f} ítems")
print("Soporte de cada ítem:"); display(M.mean().sort_values(ascending=False).round(3).to_frame("soporte").T)
print("Tres canastas de ejemplo:"); [print(f"   {s}: {nombre(c_)}") for s, c_ in list(zip(perfiles.index, canastas))[:3]]
```

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">A.1 · Soporte y confianza son conteos</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">La definición de la diapositiva sobre las nueve canastas, igual que en la cátedra.</p></div>

```python
def cuenta(M_, conjunto): return int(M_[list(conjunto)].all(axis=1).sum())             # en cuántas canastas aparecen todos los ítems del conjunto
def soporte(M_, conjunto): return cuenta(M_, conjunto) / len(M_)
def confianza(M_, A, B): return cuenta(M_, set(A) | set(B)) / cuenta(M_, A)               # cuenta(A ∪ B) / cuenta(A) = P(B | A)

items = list(M_han.columns)
print("Soporte de cada ítem (cuenta / 9):", {i: f"{cuenta(M_han, [i])}/9" for i in items})
pares = pd.DataFrame([{"par": nombre(p), "cuenta": cuenta(M_han, p), "soporte": soporte(M_han, p)} for p in itertools.combinations(items, 2)]).sort_values("cuenta", ascending=False)
print("Los diez pares:"); display(pares.round(3).reset_index(drop=True))
print(f"Confianza(I1 ⇒ I2) = cuenta(I1, I2) / cuenta(I1) = {cuenta(M_han, ['I1', 'I2'])}/{cuenta(M_han, ['I1'])} = {confianza(M_han, ['I1'], ['I2']):.2f} · Confianza(I2 ⇒ I1) = {cuenta(M_han, ['I1', 'I2'])}/{cuenta(M_han, ['I2'])} = {confianza(M_han, ['I2'], ['I1']):.2f}: la misma pareja, dos reglas, dos confianzas")
print("La regla del material: computadora ⇒ antivirus, soporte 2 %, confianza 60 %: el 2 % de las canastas traen las dos cosas; de quienes compran la computadora, el 60 % compra el antivirus.")
```

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">A.2 · Horizontal y vertical, y por qué no se puede enumerar</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Las dos maneras de guardar la misma base, y la cuenta de 2ᵈ − 1 conjuntos, ahora con los ítems de CAMMESA.</p></div>

```python
horizontal = {f"T{i + 1}00": set(t) for i, t in enumerate(han)}
vertical = {}
for tid, t in horizontal.items():
    for item in t: vertical.setdefault(item, set()).add(tid)                                 # un barrido: ítem → transacciones
print("Horizontal (por transacción):"); [print(f"   {tid}: {nombre(t)}") for tid, t in horizontal.items()]
print("Vertical (por ítem):"); [print(f"   {item}: {sorted(tids)}") for item, tids in sorted(vertical.items())]
inter = vertical["I1"] & vertical["I2"]
print(f"Lista de {{I1, I2}} = I1 ∩ I2 = {sorted(inter)} → soporte {len(inter)} (en horizontal, contando: {cuenta(M_han, ['I1', 'I2'])})")
for d in [5, 12, 20, 50, 100]: print(f"d = {d:3d} ítems → 2^d − 1 = {2 ** d - 1:.3g} conjuntos posibles")
print(f"Contar el soporte de los {2 ** M.shape[1] - 1:,} conjuntos de los {M.shape[1]} ítems de CAMMESA contra {len(M)} canastas, uno por uno: {(2 ** M.shape[1] - 1) * len(M):,} comparaciones; con 100 ítems no alcanza la edad del universo")
```

<div style="background: linear-gradient(90deg, #0055a5 0%, #003366 100%); border-radius: 8px; padding: 10px 18px; margin: 16px 0 6px 0;"><h2 style="margin: 0; color: white; font-size: 1.15em; letter-spacing: 0.5px;">B · Apriori: anti-monotonía; unir y podar por niveles</h2></div>

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">B.1 · Nivel por nivel, a mano, sobre las nueve canastas</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Barrer, unir, podar, contar; tiene que dar lo del material. Y la comparación con <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">apriori</code> de mlxtend.</p></div>

```python
def apriori_a_mano(M_, minimo, verbose=True):
    """Apriori por niveles: devuelve {conjunto: cuenta} de todos los frecuentes."""
    L = {frozenset([i]): cuenta(M_, [i]) for i in M_.columns}; L = {s: n for s, n in L.items() if n >= minimo}          # 1. barrer: L1
    if verbose: print(f"Nivel 1 · candidatos {M_.shape[1]} → frecuentes {len(L)}: " + ", ".join(f"{nombre(s)}:{n}" for s, n in sorted(L.items(), key=lambda x: sorted(x[0]))))
    frecuentes = dict(L); k = 2
    while L:
        previos = sorted(L, key=lambda s: sorted(s))
        unidos = {a | b for a, b in itertools.combinations(previos, 2) if sorted(a)[:k - 2] == sorted(b)[:k - 2] and len(a | b) == k}   # 2. unir
        candidatos = {c_ for c_ in unidos if all(frozenset(sub) in L for sub in itertools.combinations(c_, k - 1))}                     # 3. podar
        L = {c_: cuenta(M_, c_) for c_ in candidatos}; L = {s: n for s, n in L.items() if n >= minimo}                                    # 4. contar
        if verbose: print(f"Nivel {k} · unidos {len(unidos)} → podados a {len(candidatos)} candidatos → frecuentes {len(L)}" + (": " + ", ".join(f"{nombre(s)}:{n}" for s, n in sorted(L.items(), key=lambda x: sorted(x[0]))) if L else " (vacío: fin)"))
        frecuentes.update(L); k += 1
    return frecuentes

frecuentes_han = apriori_a_mano(M_han, minimo=2)
biblio = apriori(M_han, min_support=2 / 9 - 1e-9, use_colnames=True)                    # min_support es una fracción: 2 de 9 (con una tolerancia por el redondeo)
soporte_biblio = dict(zip(biblio["itemsets"], biblio["support"]))
mismos = set(soporte_biblio) == set(frecuentes_han) and all(abs(soporte_biblio[s] * 9 - n) < 1e-9 for s, n in frecuentes_han.items())
print(f"La biblioteca encuentra {len(biblio)} conjuntos frecuentes · los mismos, con los mismos soportes: {mismos}")
```

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">B.2 · De frecuentes a reglas</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Las seis reglas de {I1, I2, I5}, la poda dentro de un frecuente, y la comparación con <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">association_rules</code>.</p></div>

```python
def reglas_de(frecuentes, conjunto):
    """Todas las particiones A ⇒ B de un frecuente, con su confianza (cociente de cuentas ya hechas)."""
    filas = []
    for r in range(1, len(conjunto)):
        for A in itertools.combinations(sorted(conjunto), r):
            A = frozenset(A); B = conjunto - A
            filas.append({"regla": f"{nombre(A)} ⇒ {nombre(B)}", "cuenta(A ∪ B)": frecuentes[conjunto], "cuenta(A)": frecuentes[A], "confianza": frecuentes[conjunto] / frecuentes[A]})
    return pd.DataFrame(filas)
seis = reglas_de(frecuentes_han, frozenset(["I1", "I2", "I5"])); seis["pasa 70 %"] = seis["confianza"] >= 0.7
print("Las seis reglas de {I1, I2, I5} (cuenta 2):"); display(seis.round(2))
print("Poda dentro del frecuente: {I1, I2} ⇒ {I5} da 2/4 = 0,50 y no pasa; entonces {I1} ⇒ {I2, I5} (2/6) y {I2} ⇒ {I1, I5} (2/7) tampoco: mover un ítem del antecedente al consecuente solo puede bajar la confianza.")
todas = pd.concat([reglas_de(frecuentes_han, s) for s in frecuentes_han if len(s) >= 2], ignore_index=True)
fuertes = todas[todas["confianza"] >= 0.7]
rb = association_rules(biblio, metric="confidence", min_threshold=0.7)
print(f"Con confianza mínima 70 % sobre todos los frecuentes: {len(fuertes)} reglas fuertes a mano · {len(rb)} con association_rules · las mismas: {set(fuertes['regla']) == set(f'{nombre(a)} ⇒ {nombre(b)}' for a, b in zip(rb['antecedents'], rb['consequents']))}")
display(fuertes.sort_values("confianza", ascending=False).round(2).reset_index(drop=True))
```

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">C.1 · Eclat a mano, en quince líneas</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Sobre las nueve canastas tiene que dar lo mismo que Apriori; sobre las de CAMMESA, lo mismo que la biblioteca, con el tiempo de los cuatro. Mínimo: 17 canastas (2 %).</p></div>

```python
def eclat(canastas_, minimo):
    """Eclat: listas verticales e intersección por prefijo. Devuelve {conjunto: cuenta}."""
    listas = {}
    for tid, t in enumerate(canastas_):
        for item in t: listas.setdefault(frozenset([item]), set()).add(tid)                  # un barrido: ítem → transacciones
    listas = {s: l for s, l in listas.items() if len(l) >= minimo}
    frecuentes = {s: len(l) for s, l in listas.items()}
    def extender(prefijos):
        prefijos = sorted(prefijos, key=lambda s: sorted(s))
        for i, a in enumerate(prefijos):
            nuevos = {}
            for b in prefijos[i + 1:]:
                if sorted(a)[:-1] == sorted(b)[:-1]:                                          # mismo prefijo: comparten todo salvo el último ítem
                    inter = listas[a] & listas[b]                                             # la lista del candidato: la intersección
                    if len(inter) >= minimo: listas[a | b] = inter; nuevos[a | b] = len(inter)
            if nuevos: frecuentes.update(nuevos); extender(list(nuevos))                     # si llega, se guarda y se extiende; si no, se abandona la rama
    extender(list(listas)); return frecuentes

ec_han = eclat(han, 2)
print(f"Nueve canastas, mínimo 2 · Eclat: {len(ec_han)} frecuentes · iguales a los de Apriori a mano (conjuntos y cuentas): {ec_han == frecuentes_han}")
t0 = time.perf_counter(); ec = eclat(canastas, 17); t_e = time.perf_counter() - t0
t0 = time.perf_counter(); ap_ = apriori_a_mano(M, 17, verbose=False); t_m = time.perf_counter() - t0
t0 = time.perf_counter(); fb = apriori(M, min_support=17 / len(M) - 1e-9, use_colnames=True); t_b = time.perf_counter() - t0
t0 = time.perf_counter(); fg = fpgrowth(M, min_support=17 / len(M) - 1e-9, use_colnames=True); t_g = time.perf_counter() - t0
print(f"Canastas de CAMMESA, mínimo 17 (2 %) · Eclat a mano {len(ec)} frecuentes en {t_e * 1000:.0f} ms · Apriori a mano {len(ap_)} en {t_m * 1000:.0f} ms · apriori (mlxtend) {len(fb)} en {t_b * 1000:.0f} ms · fpgrowth {len(fg)} en {t_g * 1000:.0f} ms · todos los mismos conjuntos: {set(ec) == set(ap_) == set(fb['itemsets']) == set(fg['itemsets'])}")
print("El resultado no cambia; cambia la estrategia: Apriori barre la base una vez por nivel y cuenta subconjuntos; Eclat la lee una vez y después intersecta listas en memoria; FP-growth no genera candidatos.")
```

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">D.1 · Fuerte no es interesante: juegos y videos</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">El ejemplo de Han, con la tabla de contingencia, la elevación, la convicción y el apalancamiento, igual que en la cátedra.</p></div>

```python
N, j, v, jv = 10_000, 6000, 7500, 4000
tabla_c = pd.DataFrame([[jv, j - jv, j], [v - jv, N - j - v + jv, N - j], [v, N - v, N]], index=["juego", "no juego", "total"], columns=["video", "no video", "total"])
print("La tabla de contingencia del material:"); display(tabla_c)
sop, conf = jv / N, jv / j; p_video = v / N
elevacion = sop / ((j / N) * (v / N)); conviccion = (1 - p_video) / (1 - conf); apalancamiento = sop - (j / N) * (v / N)
print(f"juego ⇒ video: soporte {sop:.2f} · confianza P(video | juego) = {conf:.3f} · pero P(video) = {p_video:.2f}: comprar el juego BAJA la chance de comprar el video")
print(f"Elevación = P(juego ∪ video) / (P(juego) · P(video)) = {sop:.2f} / ({j / N:.2f} · {v / N:.2f}) = {elevacion:.3f} (< 1: correlación negativa) · = confianza / soporte(video) = {conf:.3f} / {p_video:.2f} = {conf / p_video:.3f}")
print(f"Convicción = (1 − soporte(video)) / (1 − confianza) = {1 - p_video:.2f} / {1 - conf:.3f} = {conviccion:.3f} (< 1: la regla falla MÁS de lo que fallaría por independencia) · apalancamiento = {sop:.2f} − {j / N:.2f} · {v / N:.2f} = {apalancamiento:+.3f}")
# comprobación con la biblioteca, construyendo las 10 000 transacciones
tr = [["juego", "video"]] * jv + [["juego"]] * (j - jv) + [["video"]] * (v - jv) + [["nada"]] * (N - j - v + jv)
M_jv = pd.DataFrame(codificador.fit(tr).transform(tr), columns=codificador.columns_)
r_jv = association_rules(apriori(M_jv, min_support=0.3, use_colnames=True), metric="confidence", min_threshold=0.5)
r_jv = r_jv[(r_jv["antecedents"] == frozenset(["juego"])) & (r_jv["consequents"] == frozenset(["video"]))]
print(f"La biblioteca, con las 10 000 transacciones armadas: confianza {float(r_jv['confidence'].iloc[0]):.3f} · elevación {float(r_jv['lift'].iloc[0]):.3f} · convicción {float(r_jv['conviction'].iloc[0]):.3f} · apalancamiento {float(r_jv['leverage'].iloc[0]):+.3f}")
```

Las de mayor elevación entre ítems sueltos: arriba de todo, una pareja <b>obvia por construcción</b> (volátil con caídas frecuentes, que salen de la misma serie). Las de mayor confianza (1,0): <b>definiciones</b> («tarifa residencial ⇒ distribuidor», «gran usuario ⇒ tarifa industrial») con otros ítems de acompañantes. Y las <b>interesantes</b>: un atributo (región, tipo, tarifa) de antecedente y un comportamiento de consecuente, con soporte mínimo 1 %. Al final, las negativas.</p></div>

```python
frec = apriori(M, min_support=0.02, use_colnames=True)
reglas = association_rules(frec, metric="lift", min_threshold=1.0)
reglas["regla"] = [f"{nombre(a)} ⇒ {nombre(b)}" for a, b in zip(reglas["antecedents"], reglas["consequents"])]
columnas = ["regla", "support", "confidence", "lift", "leverage", "conviction"]
print(f"{len(frec)} frecuentes → {len(reglas)} reglas con elevación > 1 (de {len(association_rules(frec, metric='confidence', min_threshold=0.0))} posibles)")
print("1 · Las ocho de mayor elevación, con un ítem de cada lado (la primera pareja es obvia por construcción):")
display(reglas[(reglas["antecedents"].apply(len) == 1) & (reglas["consequents"].apply(len) == 1)].sort_values("lift", ascending=False)[columnas].head(8).round(3).reset_index(drop=True))
print("2 · Las cinco de mayor confianza (confianza 1: son definiciones, como «residencial ⇒ distribuidor» o «gran usuario ⇒ tarifa industrial», con otros ítems de acompañantes):")
display(reglas.sort_values("confidence", ascending=False)[columnas].head(5).round(3).reset_index(drop=True))

es_descriptivo = lambda s: all(i in DESCRIPTIVOS for i in s)
es_comportamiento = lambda s: all(i in COMPORTAMIENTO for i in s)
frec1 = apriori(M, min_support=0.01, use_colnames=True)
cruzadas = association_rules(frec1, metric="confidence", min_threshold=0.0)
cruzadas = cruzadas[cruzadas["antecedents"].apply(es_descriptivo) & cruzadas["consequents"].apply(es_comportamiento) & (cruzadas["consequents"].apply(len) == 1)].copy()
cruzadas["regla"] = [f"{nombre(a)} ⇒ {nombre(b)}" for a, b in zip(cruzadas["antecedents"], cruzadas["consequents"])]
# entre reglas con la misma conclusión y el mismo conteo, se queda la de antecedente más corto (no repetir «distribuidor» cuando no agrega)
cruzadas = cruzadas.sort_values(["consequents", "support", "lift"], ascending=[True, False, False])
cruzadas["largo"] = cruzadas["antecedents"].apply(len)
cruzadas = cruzadas.sort_values("largo").drop_duplicates(subset=["consequents", "support", "confidence"])
print("3 · Las interesantes: un atributo ⇒ un comportamiento, las doce de mayor elevación (soporte ≥ 1 %):")
display(cruzadas.sort_values("lift", ascending=False)[columnas].head(12).round(3).reset_index(drop=True))
print("4 · Las negativas: un atributo que hace MENOS probable un comportamiento (elevación < 1, soporte del antecedente ≥ 5 %):")
negativas = cruzadas[(cruzadas["antecedent support"] >= 0.05)].sort_values("lift")
display(negativas[columnas].head(6).round(3).reset_index(drop=True))

todas_reglas = association_rules(frec, metric="confidence", min_threshold=0.0)
obvia = todas_reglas.apply(lambda r: set(r["antecedents"] | r["consequents"]) >= {"volátil", "caídas frecuentes"} or (r["consequents"] == frozenset(["tarifa industrial"])) or (r["consequents"] == frozenset(["gran usuario"])), axis=1)
cruzada = todas_reglas["antecedents"].apply(es_descriptivo) & todas_reglas["consequents"].apply(es_comportamiento)
fig, ax = plt.subplots(figsize=(10, 4.8))
ax.scatter(todas_reglas.loc[~obvia & ~cruzada, "confidence"], todas_reglas.loc[~obvia & ~cruzada, "lift"], s=10, alpha=0.4, color=GRIS, label="otras reglas")
ax.scatter(todas_reglas.loc[obvia, "confidence"], todas_reglas.loc[obvia, "lift"], s=18, color=ROJO, marker="s", alpha=0.6, label="obvias: volátil con caídas, o hacia «tarifa industrial» / «gran usuario»")
ax.scatter(todas_reglas.loc[cruzada, "confidence"], todas_reglas.loc[cruzada, "lift"], s=26, color=VERDE, alpha=0.9, label="un atributo ⇒ un comportamiento")
ax.axhline(1, color=GRIS, ls=":"); ax.set_xlabel("confianza"); ax.set_ylabel("elevación"); ax.legend(fontsize=8)
ax.set_title(f"Las {len(todas_reglas)} reglas de los frecuentes (soporte ≥ 2 %): las obvias dominan arriba a la derecha; las útiles son las verdes", fontsize=10)
plt.show()
```

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">Qué se ve</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Las reglas con más elevación y más confianza incluyen a las que menos enseñan: «volátil» y «caídas frecuentes» se miden sobre los mismos cambios mensuales, y las de confianza 1 son definiciones del tipo de agente o la tarifa. Son la trampa de D.1 en estos datos: fuertes y sin información.</p><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Las verdes sí dicen algo del negocio eléctrico. <div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">D.4 · Menos conjuntos, misma información: cerrados y maximales</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Sobre las nueve canastas y sobre las de CAMMESA.</p></div>

```python
def cerrados_y_maximales(frecuentes):
    cerrados = {s for s in frecuentes if not any(t > s and frecuentes[t] == frecuentes[s] for t in frecuentes)}
    maximales = {s for s in frecuentes if not any(t > s for t in frecuentes)}
    return cerrados, maximales
for nombre_, fr in [("nueve canastas (mínimo 2)", frecuentes_han), ("canastas de CAMMESA (mínimo 17)", ec)]:
    ce, ma = cerrados_y_maximales(fr)
    print(f"{nombre_}: {len(fr)} frecuentes · {len(ce)} cerrados · {len(ma)} maximales" + (f" · maximales: {', '.join(nombre(s) for s in sorted(ma, key=lambda s: sorted(s)))}" if len(ma) <= 6 else ""))
ce, ma = cerrados_y_maximales(frecuentes_han)
print("En las nueve: {I1, I2} tiene cuenta 4 y {I1, I2, I3} y {I1, I2, I5} tienen 2: {I1, I2} es cerrado (ningún superconjunto con cuenta 4) pero no maximal (tiene superconjuntos frecuentes).")
```

<div style="background: linear-gradient(90deg, #0055a5 0%, #003366 100%); border-radius: 8px; padding: 10px 18px; margin: 16px 0 6px 0;"><h2 style="margin: 0; color: white; font-size: 1.15em; letter-spacing: 0.5px;">F · Usarlo: qué esperar de un agente, y qué le falta a estos datos</h2></div>

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 12px 18px; margin: 4px 0;"><div style="margin-bottom: 6px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.9em; padding: 2px 12px; border-radius: 20px;">📖</span><b style="color: #003366; margin-left: 8px;">Qué comportamiento esperar de un agente nuevo</b></div><p style="margin: 0 0 6px 0; color: #222; font-size: 0.97em; line-height: 1.55;">Lo que se hace con una regla, en código: dado lo que se sabe de un agente que recién entra al mercado (su región, si es distribuidor o gran usuario, su tarifa), las reglas cuyo antecedente está contenido en eso, ordenadas por elevación, dicen qué comportamiento esperar. Con las reglas interesantes de D.3 (un atributo ⇒ un comportamiento, elevación mayor que 1,3).</p></div>

```python
reglas_utiles = cruzadas[cruzadas["lift"] > 1.3].copy()
def esperar(atributos_agente, k=3):
    tiene = frozenset(atributos_agente)
    aplican = reglas_utiles[reglas_utiles["antecedents"].apply(lambda A: A <= tiene)]
    return aplican.sort_values("lift", ascending=False).drop_duplicates(subset=["consequents"]).head(k)[["regla", "confidence", "lift", "support"]]
for agente in [["región Noreste", "distribuidor", "tarifa residencial"], ["región Gran Bs.As.", "gran usuario", "tarifa industrial"], ["región Buenos Aires", "distribuidor", "tarifa residencial"], ["región Patagonica", "gran usuario", "tarifa industrial"]]:
    rec = esperar(agente)
    print(f"Agente {nombre(agente)}: " + (" · ".join(f"{list(r.regla.split(' ⇒ '))[1]} (confianza {r.confidence:.2f}, elevación {r.lift:.2f}, regla {r.regla.split(' ⇒ ')[0]})" for r in rec.itertuples()) if len(rec) else "ninguna regla con elevación > 1,3 aplica: no hay nada que anticipar sobre su comportamiento"))
```
